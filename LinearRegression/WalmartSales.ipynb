```python
import pandas
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from pandas import DataFrame
```

```python
data = pandas.read_csv('Walmart_Sales.csv')
```

```python
data
```

```python
X = DataFrame(data,columns=['Temperature'])
```

```python
X
```

```python
X.isnull().sum()
```

```python
data.describe()
```

```python
Y = DataFrame(data,columns=["Weekly_Sales"])
Y
```

```python
Y.isnull().sum()
```

```python
plt.figure(figsize = (10,10))
plt.title("Walmart region temperature VS Weekly Sales")
plt.xlabel("Temperature")
plt.ylabel("Weekly_Sales")
plt.scatter(X,Y,alpha=0.3)
plt.ylim(0)
plt.show()
```

```python
regression = LinearRegression()
regression.fit(X,Y)
```

```python
plt.figure(figsize = (10,10))
plt.title("Walmart region temperature VS Weekly Sales")
plt.xlabel("Temperature")
plt.ylabel("Weekly Sales")
plt.scatter(X,Y,alpha=0.3)
plt.plot(X,regression.predict(X),color='red',linewidth=5)
plt.ylim(0)
plt.show()
```

```python
regression.intercept_
```

```python
regression.coef_
```

```python
regression.score(X,Y)
```

```python
# Conclusion: Temperature & weekly sales does not have linear relationship between them. Linear Regression does not fit into this & has very bad R score
```
