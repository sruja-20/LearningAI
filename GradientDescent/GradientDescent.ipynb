# Import packages

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

# Example 1
## $f(x)=x^2+x+1$

```python
def f(x):
    return x**2 + x + 1
```

```python
# make data
x_1 = np.linspace(start = -3,stop = 3,num=100)
```

```python
plt.xlim([-3,3])
plt.ylim([0,8])
plt.xlabel('x',fontsize=16)
plt.ylabel('f(x)',fontsize=16)
plt.plot(x_1,f(x_1))
plt.show()
```

# slope & derivative

```python
def df(x):
    return 2*x + 1
```

```python
plt.figure(figsize=[10,5])
plt.subplot(1,2,1)
plt.xlim([-3,3])
plt.ylim([0,8])
plt.xlabel('x',fontsize=16)
plt.ylabel('f(x)',fontsize=16)
plt.plot(x_1,f(x_1),color='blue',linewidth=3)
plt.subplot(1,2,2)
plt.xlim([-2,3])
plt.ylim([-2,6])
plt.xlabel('x',fontsize=16)
plt.ylabel('df(x)',fontsize=16)
plt.plot(x_1,df(x_1),color="skyblue",linewidth=5)
plt.grid()
plt.show()
plt.show()
```

## Gradient decent algorithm

```python
new_x = 3
previous_x = 0
learning_rate = 0.1
new_x_list = [new_x]
slope = [df(new_x)]
for i in range(100):
    previous_x = new_x
    gradient = df(new_x)
    new_x = new_x - learning_rate * gradient
    new_x_list.append(new_x)
    slope.append(df(new_x))
print(new_x)
print(df(new_x))
```

```python
plt.figure(figsize=[15,5])
plt.subplot(1,3,1)
plt.xlim([-3,3])
plt.ylim([0,8])
plt.xlabel('x',fontsize=16)
plt.ylabel('f(x)',fontsize=16)
plt.plot(x_1,f(x_1),color='blue',linewidth=3)
new_x_list = np.array(new_x_list)
plt.scatter(new_x_list,f(new_x_list),color='red',s=200,alpha=0.5)
plt.subplot(1,3,2)
plt.xlim([-2,3])
plt.ylim([-2,6])
plt.xlabel('x',fontsize=16)
plt.ylabel('df(x)',fontsize=16)
plt.plot(x_1,df(x_1),color="skyblue",linewidth=5)
plt.scatter(new_x_list,slope,color='red',s=200,alpha=0.5)
plt.grid()
plt.subplot(1,3,3)
plt.xlim([-1,1])
plt.ylim([-2,6])
plt.xlabel('x',fontsize=16)
plt.ylabel('df(x)',fontsize=16)
plt.plot(x_1,df(x_1),color="skyblue",linewidth=5)
plt.scatter(new_x_list,slope,color='red',s=200,alpha=0.5)
plt.grid()
plt.show()
```

# Example 2

## $$g(x) = x^4 - 4x + 5$$

## $$g(x) = x^4 - 4x + 5$$

## $$g(x) = x^4 -4x^2 + 5 $$

```python
# generate data
x_2 = np.linspace(-2,2,1000)
```

```python
def g(x):
    return x**4 - 4*x**2 + 5
```

```python
def dg(x):
    return 4*x**3 - 8*x
```

```python
plt.figure(figsize=[10,5])
plt.subplot(1,2,1)
plt.xlim([-2,2])
plt.ylim([0.5,6])
plt.xlabel('x_2',fontsize=16)
plt.ylabel('g(x)',fontsize=16)
plt.plot(x_2,g(x_2),color='blue',linewidth=3)
plt.subplot(1,2,2)
plt.xlim([-2,2])
plt.ylim([-6,8])
plt.xlabel('x_2',fontsize=16)
plt.ylabel('dg(x)',fontsize=16)
plt.plot(x_2,dg(x_2),color="skyblue",linewidth=5)
plt.grid()
plt.show()
plt.show()
```

```python
def gradientDecent(derivative_fun,initial_guess,multiplier=0.02,precision=0.001,step=500):
    new_x = initial_guess
    new_x_list = [new_x]
    slope = [derivative_fun(new_x)]
    for i in range(step):
        previous_x = new_x
        gradient = derivative_fun(new_x)
        new_x = new_x - multiplier * gradient
        new_x_list.append(new_x)
        slope.append(derivative_fun(new_x))
    return new_x,new_x_list,slope
```

```python
new_x,x_list,slope = gradientDecent(dg,0)
print("New x:",new_x)
```

```python
plt.figure(figsize=[10,5])
plt.subplot(1,2,1)
plt.xlim([-2,2])
plt.ylim([0.5,6])
plt.xlabel('x_2',fontsize=16)
plt.ylabel('g(x)',fontsize=16)
plt.plot(x_2,g(x_2),color='blue',linewidth=3,alpha=0.8)
plt.scatter(x_list,g(np.array(x_list)),s=100,alpha=0.6,c="red")
plt.subplot(1,2,2)
plt.xlim([-2,2])
plt.ylim([-6,8])
plt.xlabel('x_2',fontsize=16)
plt.ylabel('dg(x)',fontsize=16)
plt.plot(x_2,dg(x_2),color="skyblue",linewidth=5)
plt.scatter(x_list,slope,s=100,c="red")
plt.grid()
plt.show()
plt.show()
```

# Example 3 ( Divergence , overflow , pthon tuples )
## $$h(x) = x^5 - 2x^4 + 2 $$

```python

x_3 = np.linspace(-2.5,2.5,1000)

def h(x):
    return x**5 - 2*x**4 + 2
```

```python
def dh(x):
    return 5*x**4 - 8*x**3
```

```python
new_x,x_list,slope = gradientDecent(dh,-0.2,step=70)
```
