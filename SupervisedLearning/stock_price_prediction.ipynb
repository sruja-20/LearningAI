### Task 1: Linear regression model to predict close price using open price ( Write model from scratch & then compare with tensorflow neural model

```python
X = data["Open"]
Y = data["Close"]
```

```python
def compute_cost(x,y,w,b):
    return np.sum((y - (x*w + b)) ** 2)/(2*len(x))
```
